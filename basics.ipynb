```python
with open('data/shakespeare.txt') as f:
    text= f.read()
```

```python
print(len(text))
print(text[:100])
```

```python
#character mapping 
chars = sorted(list(set(text)))
stoi = { ch:i for i,ch in enumerate(chars)} #string to int dict
itos = { i:ch for i,ch in enumerate(chars)}

encode = lambda s:[stoi[c] for c in s] #transform string to ints
decode = lambda l:"".join([itos[i] for i in l])

print(encode("hi"))
print(decode(encode("hi")))
```

```python
import torch
data = torch.tensor(encode(text), dtype = torch.long )
print(data[:100])
```

```python
#split test/train
n = int(0.9*len(data))
train_data = data[:n]
val_data = data[n:]
```
